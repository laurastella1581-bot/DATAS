``` python

````python
age = 50
gen ='f'
if(age <40 and gen =='f' ):
        print("아가씨")
else : 
        print("아줌마")
````

````python
if age < 40 and gen == "f":
    print("아가씨")
else:
    print("아줌마")
````

## 멀티 조건문

* and : 둘 중 하나라도 True

````python
age = 60
gen = 'm'
print( age < 30 and gen =='m' )
````

* or : 둘 중 하나라도 True

````python
# 30미만 남성 키는 150이상
age = 25
gen = 'm'
height = 178

if ( age < 30 and gen =='m' and height>=150) :
    print("군필")
else :
    print("미필")
````

* 90이상이면A, 80이상이면B, 70이상이면C, 나머지 

````python
score = 85
if (score >=90 ) :
    print("A")
    
elif (score >=80 ) :
    print("B")

elif (score >=70 ) :
    print("C")

else:
    print("나머지")
````

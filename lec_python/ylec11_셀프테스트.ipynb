#  **제어문 : if** 

* score=90이상이면 A, 80이상이면 B, 아니면 F 출력

````python
score=100
if(score >= 90):
    print('A')
elif(score >= 80):
    print('B')
else:
    print('F')
````

#  **제어문 : while** 

* **[암기] 1부터 5까지 while문 사용해 출력**

````python
n = 1
while (True) :
    print(n, end="\t")
    if (n == 5) : 
        break
    n = n + 1
````

* **주소를 넣으면 우편번호 찾아주는 함수**
    * 함수이름 : search_post
    * 파라미터 : 주소
    * 기능 : 우편번호찾기
    * 리턴 : 우편번호

````python
def search_post(addr):
    if addr == "경기 부천시":
        return "14500"
    


a = search_post("경기 부천시")
print(a)
````

* **숫자를 입력받아 10배 값으로 반환하는 함수**
    * 함수이름 : calc_point
    * 파라미터 : num
    * 기능 : 입력값의 10배
    * 리턴 : 10배값

````python
def calc_point(n):
    return n*10
a= calc_point(5)
print(a)
````

* **숫자 2개를 입력받아 합의 값을 돌려주는 함수**
    * 함수이름 : calc_point2
    * 파라미터 : num1, num2
    * 기능 : num1 2배값과 num2 10배값의 합
    * 리턴 : 계산값

````python
def calc_point2(n1,n2):
    return n1*2 + n2*10
a=calc_point2(2,4)
print(a)
````

* **출석부에 학생이름을 입력하면 추가하는 함수**
    * 함수이름 : add_student
    * 파라미터 : name

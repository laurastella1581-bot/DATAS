# **Lec02.반복문 while**

``` python

while 조건문 :
    코드1
    코드2

````python
number = 1                  # 시작값을 1로 설정

while number <= 5:          # number가 5 이하인 동안 반복
    print(number)           # 현재 number 출력
    number = number + 1     # number에 1을 더함
````

````python
number = 1
while (True) :
    print(number)

    if (number == 5) :
        break
    number = number + 1
````

````python
d = 2                         # 2단부터 시작

while d <= 9:                 # 2단 ~ 9단까지 반복
    n = 1                       # 각 단마다 1부터 시작

    while n <= 9:               # 1 ~ 9까지 곱하기
        print(d, "*", n, "=", d * n)
        n = n + 1               # i를 1씩 증가

    d = d + 1               # 다음 단으로 이동
````

````python
age = input("나이를 입력하세요")
print(int(age) - 5)
````

````python
word = input("알파벳을 입력하세요")
if (word in ["Q", "q"] ) :
    print("종료")
else :
    print(f"입력글자 : {word}")
````

````python
while(True) : 
    word = input("알파벳을 입력하세요")
    if  ( word in ["Q", "q"] ) :
        print("종료")
        break
    else :
        print(f"입력글자 : {word}")
````

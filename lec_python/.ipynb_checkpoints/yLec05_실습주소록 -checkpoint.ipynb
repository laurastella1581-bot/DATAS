# **Lec05. [실습]주소록**

```python
addr_list = []                              # 주소록을 담을 빈 상자를 하나 만든다
                                            # 나중에 이름과 전화번호를 여기에 넣는다


while True:                                 # 계속 반복한다
                                            # Q를 눌러서 break를 만나기 전까지 계속 실행

    print("_______________________________________________")   # 구분선 출력
    print("1:등록 2:수정 3:삭제 4:조회 Q: 종료")              # 메뉴를 화면에 보여준다

    menu = input("메뉴를 선택하세요")         # 사용자가 누른 메뉴 번호를 menu에 저장한다


    if menu in ["Q", "q"]:                   # menu가 Q 또는 q라면
        print("종료")                         # "종료"라고 출력한다
        break                                # while 반복문을 완전히 끝낸다


    elif menu == "1":                        # 사용자가 1을 입력했다면
        print("등록")                         # "등록" 출력

        name = input("이름을 입력하세요")      # 입력한 이름을 name에 저장
        tel = input("전화번호를 입력하세요")    # 입력한 전화번호를 tel에 저장

        addr_list.append([name, tel])         # [이름, 전화번호]를 한 묶음으로 만들어
                                             # addr_list 안에 추가한다

        print(f"{name},{tel} 저장되었습니다")  # 입력한 이름과 전화번호가 저장됐다고 출력


    elif menu == "2":                        # 사용자가 2를 입력했다면
        print("수정")                         # "수정" 출력

        search_name = input("수정하려는 사람의 이름을 입력하세요")
                                             # 수정할 사람의 이름을 입력받아
                                             # search_name에 저장

        update_tel = input("변경될 전화번호를 입력하세요")
                                             # 새 전화번호를 입력받아
                                             # update_tel에 저장

        for v in addr_list:                  # 주소록에서 사람을 한 명씩 꺼낸다
                                             # v에는 [이름, 전화번호]가 들어간다

            if v[0] == search_name:          # v[0]은 이름이다
                                             # 그 이름이 찾는 이름과 같다면

                v[1] = update_tel            # v[1]은 전화번호다
                                             # 기존 전화번호를 새 전화번호로 바꾼다


    elif menu == "3":                        # 사용자가 3을 입력했다면
        print("삭제")                         # "삭제" 출력

        search_name = input("삭제하려는 사람의 이름을 입력하세요")
                                             # 삭제할 사람 이름 입력

        for v in addr_list:                  # 주소록에서 한 사람씩 꺼낸다

            if v[0] == search_name:          # 이름이 찾는 사람과 같다면

                addr_list.remove(v)          # 그 사람 정보 전체를 주소록에서 삭제한다


    elif menu == "4":                        # 사용자가 4를 입력했다면

        print(f"{len(addr_list)}건 조회")     # 주소록에 몇 명 있는지 출력
                                             # len() = 개수를 세는 기능

        for v in addr_list:                  # 주소록의 사람들을 한 명씩 꺼낸다

            print(v)                         # 한 사람의 [이름, 전화번호]를 출력한다


    else:                                    # 1,2,3,4,Q,q가 아닌 다른 값을 입력했다면

        print("잘못된 메뉴 번호를 선택하셨습니다.")
                                             # 잘못 입력했다고 알려준다
```

## print ( 값 or 변수 , end="\n")

````python
for v in range(1, 11):
    print(v * 10 , end="")
    if(v != 10):
        print("_", end="")
    else : 
        print("", end="")
````

````python
for v in range(1,6) :
    print(v * 10, end="")
    if(v == 5) :
        print("", end="")
    else :
        print("-", end="")
````

* range사용
* 1 ~ 10 짝수면 값짝, 홀수면 값홀 출력
* 줄바꿈없이

````python
for v in range(1, 11):          
    if v % 2 == 0:               
        print(v,"짝", end="수다")   
    else:                        
        print(v,"홀", end="수다")  
````

* 1부터 10까지 합

````python
total = 0
for number in range(1,11) :
     
    total = total + number

print(total)        
````

* 1부터 10까지 짝수 합

````python
total = 0
for v in range(1,11) :
    if v % 2 == 0:
         
     
        total = total + v

print(total)        
````

* 리스트에서 20 보다 작은 3의 배수를 출력하라
* mylist = [13, 21, 12, 14, 30, 18]

````python
mylist = [13, 21, 12, 14, 30, 18]

for v in mylist: 
    if v < 20 and v % 3 == 0:
        print (v)
````

* 구구단

````python
for dan in range(2,10) : #2 ~ 9
    #print(dan, '단')
    print(f"{dan}단")
````

````python
for gob in range(1,10) :
    print(f"*{gob}=", end="\t")
````

````python
{dan}단
    {dan}*{gob}={dan*gob}
    {dan}*{gob}=4
    {dan}*{gob}=6
    ...
    {dan}*{gob}=18
--------------------------
{dan}단
    {dan}*1=2
    {dan}*2=4
    {dan}*3=6
    
    
    
````

````python
for d in range(2, 10):
    print(f"{d}단" , end="\n")
    for v in range(1, 10):           
        print(d, "x",  v, "=", d * v , end="\n") 
        
````

````python
for d in range(2, 10):         
    print(f"{d}단", end="\t")

print() 
                         

for g in range(1, 10):
    for d in range(2, 10):      
        print(f"{d}x{g}={d*g}", end="\t")
    print()                    
````

# **컨프리헨션 & 람다**

* <b>for문 1~10 짝수만 출력

```python
for n in range(1,11) :
    if  (n%2 == 0) :
        print(n)
```

```python
[n for n in range(1,11) if  (n%2 == 0) ]
```

* <b>for문 1~5 3보다 크면 값*2, 그렇지 않으면 값 출력 

```python
for n in range(1,6) :
    if  (n>3) :
        print(n*2)
    else:
        print(n)
```

```python
[n*2 if(n>3) else n for n  in range(1,6)]
```

# **람다(Lamda)**
* 익명(이름없는) 함수
* 1회용

* <b>함수이름 : add 
* 파라미터 : 숫자2개
* 합산 결과 리턴

```python
def add(a, b) : 
    res = a + b
    return res
```

```python
a = add(1,2)
print(a)
```

```python
def ncheck(n) : 
    print(n*2)

ncheck(6)
```

```python
myf = lambda n : n*2
```

```python
myf(6)
```

```python
ncheck(6)
```

* <b>람다 예제

```python
def add(a, b) : 
    res = a + b
    return res

a=add(1, 2)
print(a)
```

```python
add2 = lambda a, b : a + b

a= add2(1 ,2)
print(a)
```
